# 00 - Supabase Connection Smoke Test

Loads Supabase credentials from `app.env` and runs quick table reads.

```python
from pathlib import Path
import sys

root = Path.cwd().resolve()
for candidate in [root, *root.parents]:
    if (candidate / 'analytics' / 'notebook_utils.py').exists():
        repo_root = candidate
        break
else:
    raise RuntimeError('Could not locate repo root containing analytics/notebook_utils.py')

sys.path.insert(0, str(repo_root / 'analytics'))
import notebook_utils as nb

client = nb.get_client()
env = nb.load_env()
url = (env.get('SUPABASE_URL') or f"https://{env.get('SUPABASE_PROJECT_ID', '').strip()}.supabase.co").rstrip('/')
print('Connected client for:', url)
```

```python
tables = [
    'gift_events',
    'competition_events',
    'battle_results',
    'play_events',
]

for table in tables:
    print(f'\n=== {table} ===')
    try:
        df = nb.table_df(client, table, limit=5)
        display(df)
    except Exception as exc:
        print('Query failed:', exc)
```
